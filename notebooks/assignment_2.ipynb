```python
# Import Libraries

import numpy as np 
import pandas as pd
import re

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
# Load dataset
df = pd.read_csv("data/movie_reviews.csv")
df.head()
```

```python
df.info()
df["sentiment"].count()
```

```python
# Set random seed
RANDOM_SEED = 911
```

```python
# Split data in training and testing sets
X_train, X_test, y_train, y_test = train_test_split(
    df["review"],
    df["sentiment"],
    test_size=0.2,
    random_state=RANDOM_SEED,
    stratify=df["sentiment"]
)
```

```python
# Preprocess text
def clean_text(text):
    text = text.lower()
    text = re.sub(r"[^a-z\s]", "", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text
```

```python
X_train_clean = X_train.apply(clean_text)
X_test_clean = X_test.apply(clean_text)
```

---
### Feature Extraction 1 : Bag of Words

---

```python
# Feature extraction 1 : Bag of Words 
bow = CountVectorizer(
    stop_words="english",
    max_features=5000
)

X_train_bow = bow.fit_transform(X_train_clean)
X_test_bow = bow.transform(X_test_clean)
```

```python
# Train a model on BoW 
model_bow = LogisticRegression(max_iter=1000)
model_bow.fit(X_train_bow, y_train)

y_pred_bow = model_bow.predict(X_test_bow)
```

```python
# Evaluate BoW code
print("BoW Accuracy:", accuracy_score(y_test, y_pred_bow))
print(confusion_matrix(y_test, y_pred_bow))
print(classification_report(y_test, y_pred_bow))
```

---
### Feature Extraction 2 : TF-IDF Unigrams

---

```python
# Feature Extraction 2 : TF-IDF (Unigrams)
tfidf_uni = TfidfVectorizer(
    stop_words="english",
    max_features=5000
)

X_train_tfidf_uni = tfidf_uni.fit_transform(X_train_clean)
X_test_tfidf_uni = tfidf_uni.transform(X_test_clean)
```

```python
# Train and evaluate tf-idf (unigrams)
model_tfidf_uni = LogisticRegression(max_iter=1000)
model_tfidf_uni.fit(X_train_tfidf_uni, y_train)

y_pred_tfidf_uni = model_tfidf_uni.predict(X_test_tfidf_uni)

print("TF_IDF (Unigram) Accuracy: ",accuracy_score(y_test, y_pred_tfidf_uni))
```

---
### Feature Extraction 3 : TF-IDF Bigrams

---

```python
# Feature Extraction 3: TF_IDF (BIGRAMS)
tfidf_bi = TfidfVectorizer(
    stop_words="english",
    max_features=7000,
    ngram_range=(1, 2)
)

X_train_tfidf_bi = tfidf_bi.fit_transform(X_train_clean)
X_test_tfidf_bi = tfidf_bi.transform(X_test_clean)
```

```python
# Train and Evaluate Bigram model
model_tfidf_bi = LogisticRegression(max_iter=1000)
model_tfidf_bi.fit(X_train_tfidf_bi, y_train)

y_pred_tfidf_bi = model_tfidf_bi.predict(X_test_tfidf_bi)

print("TF-IDF Accuracy:", accuracy_score(y_test, y_pred_tfidf_bi))
```

---
### Comparison Table

---

```python
# Compare all feature extractions
results = pd.DataFrame({
    "Feature Extraction": ["BoW", "TF-IDF (Unigram)", "TF-IDF (Unigram + Bigram)"],
    "Accuracy": [
        accuracy_score(y_test, y_pred_bow),
        accuracy_score(y_test, y_pred_tfidf_uni),
        accuracy_score(y_test, y_pred_tfidf_bi)
    ]
})

results
```

### Conclusion :

In this experiment, feature extraction techniques were implemented to convert unstructured text data into numerical representations suitable for machine learning models. Both Bag-of-Words and TF-IDF vectorization methods were explored, with TF-IDF producing better performance by assigning higher importance to informative words while reducing the impact of common terms.

After feature extraction, a classification model was trained to predict sentiment polarity (positive or negative), achieving acceptable accuracy. The results clearly demonstrate that effective feature extraction plays a crucial role in text-based machine learning tasks.

This experiment highlights that raw text cannot be directly processed by ML algorithms and must be transformed into meaningful features. Proper preprocessing and vectorization significantly improve model performance, making feature engineering a critical step in natural language processing applications such as sentiment analysis.
